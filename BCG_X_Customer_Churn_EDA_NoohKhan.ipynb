```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

client_df = pd.read_csv("client_data (1).csv")
price_df = pd.read_csv("price_data (1).csv")
```

# Data Visualization

```python
plt.figure(figsize=(6,4))
sns.countplot(x='churn', data=client_df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(client_df['cons_12m'], bins=30)
plt.title("Electricity Consumption Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(client_df['num_years_antig'], bins=20)
plt.title("Customer Tenure Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(client_df['net_margin'], bins=30)
plt.title("Net Margin Distribution")
plt.show()
```

```python
plt.figure(figsize=(5,4))
sns.countplot(x='has_gas', data=client_df)
plt.title("Gas Customer Distribution")
plt.show()
```

# Key Findings

• The datasets contain customer information, pricing history, and churn indicators.

• Consumption and tenure variables show significant variation among customers.

• Pricing may influence churn, but other business factors should also be considered.

• Further feature engineering and predictive modeling can help identify the strongest churn drivers.
